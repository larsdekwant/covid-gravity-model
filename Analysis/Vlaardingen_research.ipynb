```python
# path_mezuro = ("../Mezuro_data/nederland/gemeente niveau/per dag/Bestemmingen bewoners per dag - van gemeente naar gemeente - 01-03-2019 tm 14-03-2019 - ilionx.csv")
path_gem = ("../Data/CBS/gemeente_2019/gemeente_populatie_2019.csv")
path_gem_naam = ("../Data/CBS/gemeente_2019/Gemeenten alfabetisch 2019.xls")
path_demo = ("../Data/DemographicMatrix.csv")

df_Gem = pd.read_csv(path_gem, delimiter=';', encoding='latin-1')
df_Gem_naam = pd.read_excel(path_gem_naam)
df_Gem = df_Gem.merge(df_Gem_naam, left_on='RegioS', right_on='GemeentecodeGM')
UniLocs, reverseloc = np.unique(df_Gem.Gemeentenaam, return_index=True)
```

```python
def find_mobility(gem1, gem2, mobmat):
    gem1id = df_Gem[df_Gem.Gemeentenaam == gem1].index[0]
    gem2id = df_Gem[df_Gem.Gemeentenaam == gem2].index[0]
    return mobmat[gem1id, gem2id]
```

```python
homegem = "Vlaardingen"
destgem = "Rotterdam"

mez = find_mobility(homegem, destgem, mez_freq)
grav = find_mobility(homegem, destgem, grav_freq)
grav2 = find_mobility(homegem, destgem, grav2_freq)
grav3 = find_mobility(homegem, destgem, grav3_freq)

reversemez = find_mobility(destgem, homegem, mez_freq)
reversegrav = find_mobility(destgem, homegem, grav_freq)
reversegrav2 = find_mobility(destgem, homegem, grav2_freq)
reversegrav3 = find_mobility(destgem, homegem, grav3_freq)
```

```python
print(f"Mezuro: {mez}")
print(f"Gravity: {grav}")
print(f"Gravity2: {grav2}")
print(f"Gravity3: {grav3}")

print(f"Mezuro reverse: {reversemez}")
print(f"Gravity reverse: {reversegrav}")
print(f"Gravity2 reverse: {reversegrav2}")
print(f"Gravity3 reverse: {reversegrav3}")
```
